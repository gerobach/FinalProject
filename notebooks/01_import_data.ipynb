```python
import pandas as pd

df = pd.read_csv("../data/processed/analytical_dataset.csv")
```

```python
df.columns
```

```python
df.rename(columns={'Triglycerides (mmol/L)_y': 'Triglycerides (mmol/L)'}, inplace=True)

```

```python
df.columns = (
    df.columns
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
)
df.columns
```

```python
df.drop(columns=["triglycerides_mmol/l_x"], inplace=True)
```

```python
df.rename(columns={'ever_told_diabetes?': 'diabetes?'}, inplace=True)
df.rename(columns={'ever_told_pre-diabetes?': 'pre_diabetes?'}, inplace=True)
df.rename(columns={'ever_told_high_bp?': 'high_bp?'}, inplace=True)
```

```python
df.drop(columns=["days_+4_drinks", "days_+8_drinks?"], inplace=True)
```

```python
df["haemoglobin_mmol/l"] = df["haemoglobin_g/dl"] / 1.611
```

```python
df.drop(columns=["haemoglobin_g/dl"], inplace=True)
df.insert(
    loc=14,
    column="haemoglobin_mmol/l",
    value=df.pop("haemoglobin_mmol/l")
)
```

```python
df.drop(columns=["age_diabetes"], inplace=True)
```

```python
#create new columns for waist-hip ratio and neutrophil-lymphocyte ratio

df["waist_hip_ratio"] = (
    df["waist_circumference_cm"] /
    df["hip_circumference_cm"]
)

df.insert(
    loc=9,
    column="waist_hip_ratio",
    value=df.pop("waist_hip_ratio")
)

df["nlr"] = (
    df["neutrophils_abs."] /
    df["lymphocytes_abs."]
)

df.insert(
    loc=13,
    column="nlr",
    value=df.pop("nlr")
)

df["non_hdl_mmol_l"] = (
    df["total_cholesterol_mmol/l"] -
    df["hdl_mmol/l"]
)
```

```python
df.columns
```

```python
df = df[df["age"] >= 18]
df = df.reset_index(drop=True)
```
