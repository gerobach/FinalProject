```python
questionnaire_cols = [

    "alc_consum?",

    "smoked_100_cigs?",
    "smoker_now?",

    "high_bp?",
    "bp_medication?",

    "high_cholesterol?",
    "cholesterol_medication?",

    "diabetes?",
    "pre_diabetes?",
    "insulin?",
    "diabetes_medication?"

]

missing_codes = [
    7, 9,
    77, 99,
    777, 999
]

for col in questionnaire_cols:
    df[col] = df[col].replace(missing_codes, pd.NA)
    
```

```python
df[questionnaire_cols].describe(include="all").T
```

```python
for col in questionnaire_cols:
    df[col] = df[col].astype("category")


df[questionnaire_cols].describe(include="all")
```

```python
for col in questionnaire_cols:
    print(col)
    print(df[col].unique())
    print()
```

```python
df["id"] = df["id"].astype("int64")
df["gender"] = df["gender"].astype("category")
df["ethnicity"] = df["ethnicity"].astype("category")


df["avg_drink_day_12m"] = (df["avg_drink_day_12m"].astype("float64"))

df.columns = (
    df.columns
    .str.lower()
    .str.replace("/", "_", regex=False)
    .str.replace("%", "_percent", regex=False)
    .str.replace("μ", "u", regex=False)
    .str.replace("?", "", regex=False)
    .str.replace(".", "", regex=False)
)
```
